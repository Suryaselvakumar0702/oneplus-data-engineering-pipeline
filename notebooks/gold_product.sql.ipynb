```sql
-- ============================================================
-- GOLD LAYER: PRODUCT-LEVEL BUSINESS SUMMARY
-- Source: silver_oneplus_products
-- Purpose: Product-level pricing, SKU and stock analysis
-- ============================================================

CREATE OR REPLACE TABLE workspace.default.gold_oneplus_product_summary
USING DELTA

COMMENT 'Gold layer: Product-level OnePlus product and SKU summary'

AS

SELECT

    category_name,
    product_name,
    product_code,

    COUNT(DISTINCT sku_code) AS sku_count,

    ROUND(MIN(now_price), 2) AS minimum_price,

    ROUND(MAX(now_price), 2) AS maximum_price,

    ROUND(AVG(now_price), 2) AS average_price,

    ROUND(AVG(discount_amount), 2) AS average_discount,

    COUNT(
        CASE
            WHEN stock_status = 'In Stock' THEN 1
        END
    ) AS in_stock_skus,

    COUNT(
        CASE
            WHEN stock_status = 'Out of Stock' THEN 1
        END
    ) AS out_of_stock_skus,

    COUNT(
        CASE
            WHEN stock_status = 'Low Stock' THEN 1
        END
    ) AS low_stock_skus,

    COUNT(
        CASE
            WHEN data_quality_flag = 'VALID' THEN 1
        END
    ) AS valid_records,

    COUNT(
        CASE
            WHEN data_quality_flag != 'VALID' THEN 1
        END
    ) AS flagged_records

FROM workspace.default.silver_oneplus_products

WHERE data_quality_flag = 'VALID'

GROUP BY
    category_name,
    product_name,
    product_code;


SELECT *
FROM workspace.default.gold_oneplus_product_summary;



-- ============================================================
-- GOLD LAYER: CATEGORY-LEVEL BUSINESS SUMMARY
-- ============================================================

CREATE OR REPLACE TABLE workspace.default.gold_oneplus_category_summary
USING DELTA

COMMENT 'Gold layer: Category-level OnePlus product analytics'

AS

SELECT

    category_name,

    COUNT(DISTINCT product_code) AS product_count,

    COUNT(DISTINCT sku_code) AS sku_count,

    ROUND(AVG(now_price), 2) AS average_price,

    ROUND(MIN(now_price), 2) AS minimum_price,

    ROUND(MAX(now_price), 2) AS maximum_price,

    ROUND(AVG(discount_amount), 2) AS average_discount,

    COUNT(
        CASE
            WHEN stock_status = 'In Stock' THEN 1
        END
    ) AS in_stock_skus,

    COUNT(
        CASE
            WHEN stock_status = 'Out of Stock' THEN 1
        END
    ) AS out_of_stock_skus,

    COUNT(
        CASE
            WHEN stock_status = 'Low Stock' THEN 1
        END
    ) AS low_stock_skus

FROM workspace.default.silver_oneplus_products

WHERE data_quality_flag = 'VALID'

GROUP BY category_name;



-- ============================================================
-- GOLD LAYER: STOCK ANALYSIS
-- ============================================================

CREATE OR REPLACE TABLE workspace.default.gold_oneplus_stock_summary
USING DELTA

COMMENT 'Gold layer: OnePlus product stock availability analysis'

AS

SELECT

    category_name,

    stock_status,

    COUNT(DISTINCT product_code) AS product_count,

    COUNT(DISTINCT sku_code) AS sku_count

FROM workspace.default.silver_oneplus_products

WHERE data_quality_flag = 'VALID'

GROUP BY
    category_name,
    stock_status;




CREATE OR REPLACE TABLE workspace.default.gold_oneplus_price_analysis
USING DELTA

COMMENT 'Gold layer: OnePlus pricing and discount analysis'

AS

SELECT

  category_name,

  COUNT(DISTINCT product_code) AS product_count,

  ROUND(MIN(original_price), 2)
    AS minimum_original_price,

  ROUND(MAX(original_price), 2)
    AS maximum_original_price,

  ROUND(AVG(original_price), 2)
    AS average_original_price,

  ROUND(MIN(now_price), 2)
    AS minimum_current_price,

  ROUND(MAX(now_price), 2)
    AS maximum_current_price,

  ROUND(AVG(now_price), 2)
    AS average_current_price,

  ROUND(AVG(discount_amount), 2)
    AS average_discount

FROM workspace.default.silver_oneplus_products

WHERE data_quality_flag = 'VALID'

GROUP BY category_name;
```
