```sql
-- ============================================================
-- BRONZE LAYER: RAW DATA INGESTION
-- Source: workspace.default.oneplus_all_products
-- Purpose: Ingest raw OnePlus product/SKU records
-- ============================================================

DROP TABLE IF EXISTS bronze_oneplus_products;

CREATE OR REFRESH STREAMING TABLE bronze_oneplus_products

COMMENT "Bronze layer: Raw OnePlus product and SKU records"

AS

SELECT
    *,
    current_timestamp() AS ingestion_timestamp

FROM STREAM(
    read_files(
        '/Volumes/workspace/default/oneplus/',
        format => 'csv',
        header => true,
        schemaHints => 'original_price BIGINT, sale_price BIGINT, now_price BIGINT, discount_rate BIGINT, sale_status BIGINT, sku_code BIGINT, publish_time BIGINT'
    )
);


SELECT COUNT(*) AS total_rows
FROM bronze_oneplus_products;
```
